## Question 1

**25 Points**

Given two vectors $\mathbf{x}, \mathbf{y} \in \mathbb{R}^N$, their **_cosine similarity_** is defined by

$$
  \mathrm{sim}(\mathbf{x}, \mathbf{y}) = \frac{\mathbf{x}^\top \mathbf{y}}{\lVert\mathbf{x}\rVert \lVert\mathbf{y}\rVert}
$$

Write a Python function that computes the cosine similarity of its NumPy arrays arguments.

```python
def cosine_sim(x: np.ndarray, y: np.ndarray) -> float:
    """Cosine similarity.

    Args:
        x: A NumPy array of shape (N,).
        y: A NumPy array of shape (N,).

    Returns:
        The cosine similarity of x and y.
    """
    num = np.dot(np.transpose(x), y)
    den = np.linalg.norm(x) * np.linalg.norm(y)
    return num / den
```

## Question 2

**25 Points**

Given a matrix $\mathbf{A} \in \mathbb{R}^{M \times N}$, the **Gram matrix** is defined as

$$
\mathbf{G} = \mathbf{A}^\top \mathbf{A}
$$

The Gram matrix is always **symmetric** and **positive semi-definite**.

Write a function that computes and returns $\mathbf{G}$ given $\mathbf{A}$.
Then, write a function that returns the **diagonal elements** of $\mathbf{G}$ as a 1D NumPy array.

```python
def gram_matrix(A: np.ndarray) -> np.ndarray:
    """Compute the Gram matrix of A.

    Args:
        A: A NumPy array of shape (M, N).

    Returns:
        The Gram matrix of shape (N, N).
    """
    return np.dot(np.matrix_transpose(A), A)

def gram_diagonal(A: np.ndarray) -> np.ndarray:
    """Return the diagonal of the Gram matrix of A.

    Args:
        A: A NumPy array of shape (M, N).

    Returns:
        A 1D NumPy array of length N.
    """
    M = gram_matrix(A)
    return np.diag(M)
```

Real-world data is messy. Two of the most common issues are **missing values** and **categorical variables**.
 
We provide a small dataset:

```python
df_raw = pd.DataFrame(
    {
        "temperature": [22.1, np.nan, 18.4, np.nan, 25.0],
        "weather": ["sunny", "cloudy", "rainy", "cloudy", "sunny"],
        "humidity": [0.4, 0.6, np.nan, 0.7, 0.5],
    }
)
```

Complete the following:
 
1. Assign to `n_missing` the **total number of missing values** in `df_raw` as a Python `int`.
2. Assign to `df_clean` a copy of `df_raw` where missing values in `temperature` and `humidity` are **filled with the column mean**.
3. Assign to `df_encoded` the `df_clean` DataFrame with the `weather` column replaced by **one-hot encoding** (use `pd.get_dummies`). Drop the original `weather` column.

```python
n_missing = int(df_raw.isnull().sum().sum())

df_clean = df_raw.copy()
df_clean['humidity'] = df_clean['humidity'].fillna(df_clean['humidity'].mean())
df_clean['temperature'] = df_clean['temperature'].fillna(df_clean['temperature'].mean())

df_encoded = pd.get_dummies(df_clean, columns=['weather'])
```
